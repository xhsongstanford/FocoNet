### **Load data as dictionary**

```python
import numpy as np

filepath = 'Dev/Atest_24stations.npy'
dataset = np.load(filepath, allow_pickle=True).item()
```

The keys of the dataset dictionary are the earthquake IDs

```python
eids = dataset.keys()
print(eids)
```

The value under each key contains the seismic data from each earthquake

```python
eid = list(eids)[0]
print(dataset[eid].keys())
```

The value under 'polarities' is a matrix containing the polarities, S/P ratios, SNRs.

The columns are (S/P ratios and SNRs are log values):

`polarity, max{S}/|P|, Sr/Pr, Sr/Pz, Sz/Pr, Sz/Pz, St/Pr, St/Pz, Pr/Nr, Pz/Nz, St/Nt`

The rows represent the records from each stations.

```python
print('Station Num: '+str(len(dataset[eid]['polarities'][:,0])))
print('Matrix: ')
print(dataset[eid]['polarities'])
```

The value under 'stationxyz' is a matrix including the station code, their horizontal locations, and the earthquake source depth.

The 1st column contains station codes.

The 2nd and 3rd column are each station's coordinated (X and Y) location to the hypocenter.

The 4th column is the earthquake source depth. They are the same for each row.

```python
print('Station Num: '+str(len(dataset[eid]['stationxyz'][:,0])))
print('Matrix: ')
print(dataset[eid]['stationxyz'])
```

The value under key 'focalmech' is the label focal mechanism (strike, dip, rake). It can be inaccurate or randomly set if this dataset is only used to solve for focal-mechanism-unknown earthquakes.

```python
print('strike: ' + str(dataset[eid]['focalmech'][0]))
print('dip:    ' + str(dataset[eid]['focalmech'][1]))
print('rake:   ' + str(dataset[eid]['focalmech'][2]))
```

The value under key 'mag' is the event magnitude. It can be inaccurate or randomly set.

```python
print('mag: ' + str(dataset[eid]['mag']))
```
